```python
import pandas as pd

df = pd.read_csv("geospatial_sales_dataset.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nTotal Sales:", df["Sales"].sum())
```

import pandas as pd

# Calculate total sales for each city
city_sales = df.groupby("City").agg(
    Total_Sales=("Sales", "sum"),
    Total_Orders=("Order_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Store_Presence=("Store_Presence", lambda x: (x == "Yes").any())
).reset_index()

city_sales = city_sales.sort_values(
    by="Total_Sales", ascending=False
)

print("City-wise Sales Analysis:")
display(city_sales)

print("\nCity with Highest Sales:")
display(city_sales.head(1))

```python
import plotly.express as px

# Create an interactive map
fig = px.scatter_map(
    df,
    lat="Latitude",
    lon="Longitude",
    color="Sales",
    size="Sales",
    hover_name="City",
    hover_data=["State", "Quantity", "Store_Presence"],
    color_continuous_scale="Viridis",
    zoom=3.5,
    title="Geospatial Sales Analysis Across Indian Cities",
    map_style="open-street-map"
)

fig.show()
```

```python
# Analyze business expansion opportunities

expansion = df.groupby(["City", "State", "Latitude", "Longitude"]).agg(
    Total_Sales=("Sales", "sum"),
    Total_Orders=("Order_ID", "count"),
    Existing_Stores=("Store_Presence", lambda x: (x == "Yes").sum())
).reset_index()

# Calculate the percentage of records marked as having store presence
expansion["Store_Presence_Percent"] = (
    expansion["Existing_Stores"] / expansion["Total_Orders"] * 100
).round(2)

# Rank cities by total sales
expansion = expansion.sort_values(
    by="Total_Sales", ascending=False
)

print("Business Expansion Analysis:")
display(expansion)

print("\nTop 5 Cities by Sales:")
display(expansion.head(5))
```
